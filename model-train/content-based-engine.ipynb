```python
# Import library
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import OneHotEncoder
import numpy as np
import joblib # Untuk menyimpan model/matriks

print("Memulai Fase 2A: Content-Based Engine")

# --- Langkah 2A.1: Muat Data Olahan ---
try:
    user_profile_df = pd.read_csv('../dataset/processed_user_profiles.csv')
    user_skill_matrix = pd.read_csv('../dataset/processed_user_skills.csv')
    print("Data olahan berhasil dimuat.")
except FileNotFoundError:
    print("Error: File data olahan tidak ditemukan. Jalankan notebook Fase 1 terlebih dahulu.")
    # Hentikan eksekusi jika data tidak ada
    exit()

# --- Langkah 2A.2: Feature Engineering untuk Profil ---
# Kita akan mengubah semua data profil kategorikal menjadi vektor numerik
# agar bisa dihitung kemiripannya.

# Isi nilai NaN (jika ada) dengan string 'missing' agar bisa di-encode
profile_features = user_profile_df.drop(columns=['user_id'])
profile_features = profile_features.fillna('missing')

# Gunakan OneHotEncoder untuk mengubah data kategorikal menjadi biner
encoder = OneHotEncoder(handle_unknown='ignore')
encoded_profiles = encoder.fit_transform(profile_features)

print(f"Bentuk matriks profil setelah di-encode: {encoded_profiles.shape}")

# --- Langkah 2A.3: Hitung Matriks Kemiripan Antar Pengguna ---
# Ini adalah jantung dari mesin Content-Based.
print("Menghitung matriks kemiripan pengguna (cosine similarity)...")
user_similarity_matrix = cosine_similarity(encoded_profiles)
print("Matriks kemiripan pengguna berhasil dibuat.")
print(f"Bentuk matriks kemiripan: {user_similarity_matrix.shape}")


# --- Langkah 2A.4 (REVISI): Simpan PROFIL TER-ENCODE dan Encoder ---
# Kita simpan matriks profil yang sudah di-encode, BUKAN matriks kemiripan raksasa.
joblib.dump(encoded_profiles, '../models/content_based_encoded_profiles.joblib')
joblib.dump(encoder, '../models/profile_encoder.joblib')
print("Profil ter-encode dan encoder untuk Content-Based berhasil disimpan.")

print("\n--- FASE 2A SELESAI (DENGAN PERBAIKAN) ---")
```
