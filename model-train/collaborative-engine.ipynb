```python
# Import library
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import joblib

print("Memulai Fase 2B: Collaborative-Based Engine")

# --- Langkah 2B.1: Muat Data Matriks Skill ---
try:
    user_skill_matrix = pd.read_csv('../dataset/processed_user_skills.csv')
    print("Matriks skill berhasil dimuat.")
except FileNotFoundError:
    print("Error: File matriks skill tidak ditemukan. Jalankan notebook Fase 1 terlebih dahulu.")
    exit()

# --- Langkah 2B.2: Siapkan Matriks untuk Perhitungan ---
# Kita butuh matriks skill tanpa kolom user_id
skills_only_matrix = user_skill_matrix.drop(columns=['user_id'])

# Transpose matriks agar baris menjadi skill dan kolom menjadi user
# Ini dibutuhkan untuk menghitung kemiripan antar skill
skill_user_matrix = skills_only_matrix.T
print(f"Bentuk matriks skill-user (setelah transpose): {skill_user_matrix.shape}")


# --- Langkah 2B.3: Hitung Matriks Kemiripan Antar Skill ---
# Jantung dari mesin Collaborative-Based
print("Menghitung matriks kemiripan skill (cosine similarity)...")
skill_similarity_matrix = cosine_similarity(skill_user_matrix)
print("Matriks kemiripan skill berhasil dibuat.")
print(f"Bentuk matriks kemiripan: {skill_similarity_matrix.shape}")

# --- Langkah 2B.4: Buat DataFrame dari Matriks Kemiripan ---
# Ini akan mempermudah pencarian skill nanti
skill_similarity_df = pd.DataFrame(skill_similarity_matrix,
                                   index=skills_only_matrix.columns,
                                   columns=skills_only_matrix.columns)

# --- Langkah 2B.5: Simpan Matriks Kemiripan ---
joblib.dump(skill_similarity_df, '../models/collaborative_skill_similarity.joblib')
print("Matriks kemiripan untuk Collaborative-Based berhasil disimpan.")

print("\n--- FASE 2B SELESAI ---")
```
